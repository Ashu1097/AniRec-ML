## Score spread diagnostics

```python
from src.inference.scoring_utils import zscore_normalize, sharpen_scores, power_scale

# Simulate flat anime scores (the problem we fixed)
flat_scores = np.random.normal(0.782, 0.002, 200).astype(np.float32)
print(f'Input std : {flat_scores.std():.4f}')

fixed = zscore_normalize(flat_scores)
fixed = sharpen_scores(fixed, method='tanh', strength=3.0)
fixed = power_scale(fixed, exponent=1.6)
fixed = zscore_normalize(fixed, target_lo=0.05, target_hi=0.95)
print(f'Output std: {fixed.std():.4f}')

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(flat_scores, bins=30, color='#7F77DD')
axes[0].set_title('Input (flat)')
axes[1].hist(fixed, bins=30, color='#1D9E75')
axes[1].set_title('After pipeline')
plt.tight_layout()
plt.show()
```
